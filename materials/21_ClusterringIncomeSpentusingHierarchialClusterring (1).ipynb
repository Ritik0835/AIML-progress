### Fitting the Hierarchial clustering to the dataset with n=5

```
from sklearn.cluster import AgglomerativeClustering
model = AgglomerativeClustering(n_clusters = 7, linkage ='complete')
y_means = model.fit_predict(dataset)
y_means
```

```
students = ["Rajesh","Priya","Arun","Kavi","Mani","Selvi","Ramu","Gopi","Bala","Kumar","Vikram","Deepa"]

data = np.array([
    [90, 85], [88, 92], [95, 90],
    [50, 55], [45, 60], [55, 50],
    [20, 25], [15, 30], [25, 20],
    [20, 88], [90, 20], [50, 90],
])
```

```
import numpy as np
import matplotlib.pyplot as plt
from scipy.cluster.hierarchy import dendrogram, linkage, fcluster
from sklearn.preprocessing import MinMaxScaler
scaler = MinMaxScaler()
data_scaled = scaler.fit_transform(data)
```

```
linked = linkage(data_scaled, method="ward")

# ─────────────────────────────
# Step 4: Dendrogram - tree paaru
# ─────────────────────────────
plt.figure(figsize=(10, 5))
dendrogram(linked,
           labels=students,
           leaf_rotation=45,
           leaf_font_size=11,
           color_threshold=0.8)
plt.title("Hierarchical Clustering - Student Groups")
plt.xlabel("Students")
plt.ylabel("Distance")
plt.tight_layout()
plt.show()
```
